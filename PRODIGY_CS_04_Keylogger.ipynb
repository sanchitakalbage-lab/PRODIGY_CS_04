## Task 04 - Keylogger

**Objective:**
Create a basic keylogging demonstration that records user input and saves the logged keystrokes to a file.

**Ethical Consideration:**
This project is designed for educational purposes and requires user consent. It does not capture keystrokes from other applications or run secretly in the background.

```python
log_file = "keylog.txt"

print("=== Keylogger Demonstration ===")
print("This program logs only the text you enter below.")
print("Type 'EXIT' to stop logging.\n")

with open(log_file, "a") as file:
    while True:
        user_input = input("Enter text: ")

        if user_input.upper() == "EXIT":
            break

        file.write(user_input + "\n")

print("\nLogging stopped.")
print("Keystrokes saved to:", log_file)
```

```python
with open("keylog.txt", "r") as file:
    print(file.read())
```

## Ethical Considerations

Keylogging can be used for legitimate security testing but can also violate privacy if performed without permission.

This project is designed only for educational purposes. The program records text entered directly by the user and does not monitor or capture keystrokes from other applications.

Always obtain explicit user consent before collecting or logging keystrokes.
